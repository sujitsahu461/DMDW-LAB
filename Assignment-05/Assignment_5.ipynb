```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbs
#plt.style.use('seaborn-white-grid')
```

```python
df = pd.read_csv('/content/student-mat.csv', sep=';')

df.head()
df.info()
```

```python
dfn = df[['traveltime', 'studytime']]

x = dfn['traveltime']
y = dfn['studytime']

sbs.lineplot(x=x, y=y, dashes=True)
plt.show()
```

```python
correlation = dfn.corr()

print(correlation)

sbs.heatmap(correlation, annot=True, cmap="YlGnBu")

plt.title("Correlation between Travel Time and Study Time")
plt.show()
```

```python
covariance = dfn.cov()
print(covariance)
sbs.heatmap(covariance, annot=True, cmap="YlGnBu")
plt.title("Covariance Matrix")
plt.show()
```
