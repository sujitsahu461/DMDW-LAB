```python
import pandas as pd
import numpy as np
```

```python
#for data preprocessing
from sklearn import preprocessing
```

```python
#1.create a dataset
data=pd.DataFrame({
    'age':[15,16,17,18,19,20],
    'studytime':[1,2,3,2,4,3],
    'failures':[0,1,0,2,1,0]
}).astype(float)
```

```python
#2.min max normalization
#converts values into the range[0,1]
minmax_scaler=preprocessing.MinMaxScaler()
minmax_scaled=minmax_scaler.fit_transform(data)
```

```python
#3.Z-Score Normalization
#mean=0 and standard deviation is 1
zscore_scaled=preprocessing.scale(data)
```

```python
#4.decimal scaling
j=np.ceil(np.log10(data.abs().max()))
decimal_scaled=data/(10**j)
```

```python
#Dispaly original data
print("=========ORIGINAL VALUE=========")
print(data)
```

```python
#dISPLAY MIN-MAX NORMALIZATION
print("\n====MIN-MAX NORMALIZATION=====")
print(pd.DataFrame(minmax_scaled,columns=data.columns))
```

```python
print("\n========Z-SCORE NORMALIZATION============")
print(pd.DataFrame(zscore_scaled,columns=data.columns))
```

```python
print("\n==========DECIMAL SCALING=========")
print(decimal_scaled)
```
