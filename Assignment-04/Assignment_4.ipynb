```python
import pandas as pd
import numpy as np
#Nominal dataset
data=pd.DataFrame({
    'color':['Red','Blue','Red','Green'],
    'city':['Delhi','Mumbai','Delhi','Kolkata']
})
print("NOMINAL DATASET")
print(data)
n=len(data)
matrix=np.zeros((n,n))
for i in range(n):
  for j in range(n):
    count=0
    for col in data.columns:
      if data.loc[i,col]!=data.loc[j,col]:
        count+=1
    matrix[i,j]=count/len(data.columns)
print("\n DISSIMARILTY MATRIX-NOMINAL")
print(pd.DataFrame(
    matrix,
    index=['01','02','03','04'],
    columns=['01','02','03','04']
).round(2))
```

```python
import pandas as pd
import numpy as np
#Binary dataset
data=pd.DataFrame({
    'Student':[1,1,0,0],
    'Employed':[0,1,1,0],
    'Purchased':[1,1,0,1]
})
print("BINARY DATASET")
print(data)
n=len(data)
matrix=np.zeros((n,n))
for i in range(n):
  for j in range(n):
    diff=0
    for col in data.columns:
      if data.loc[i,col]!=data.loc[j,col]:
        diff+=1
    matrix[i,j]=diff/len(data.columns)
print("\nDISSMILARITY MATRIX-BINARY")
print(pd.DataFrame(
    matrix,
   index=['01','02','03','04'],
    columns=['01','02','03','04']
).round(2))
```
